```python
from matplotlib import pyplot as plt
import pandas as pd

table = pd.read_csv("q3.txt", sep=",", header=0)

plt.scatter(table["x"], table["f(x)"], c="orange", label="1/(1+25*x²)")
plt.plot(table["x"], table["spline"], label="Spline Cúbica")
plt.legend()

plt.show()
```
